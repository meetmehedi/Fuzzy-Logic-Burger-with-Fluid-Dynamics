```python
import json, os
from IPython.display import Image, display

ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
print('Workspace root:', ROOT)

def load_strict(rel_path):
    with open(os.path.join(ROOT, rel_path)) as f:
        return json.load(f, parse_constant=lambda c: (_ for _ in ()).throw(ValueError(c)))

piv = load_strict('mewc_flow/experiments/piv_experiments/results/piv_benchmark_summary.json')
abl = load_strict('mewc_flow/experiments/piv_experiments/results/ablation_study_summary.json')
kol = load_strict('mewc_flow/experiments/kolmogorov_experiments/results/kolmogorov_benchmark_summary.json')
era = load_strict('mewc_flow/experiments/era5_experiments/results/era5_benchmark_summary.json')
brg = load_strict('mewc_flow/experiments/burgers_pilot/results/burgers_benchmark_summary.json')
print('All 5 strict-JSON benchmark summaries loaded cleanly!')
```
